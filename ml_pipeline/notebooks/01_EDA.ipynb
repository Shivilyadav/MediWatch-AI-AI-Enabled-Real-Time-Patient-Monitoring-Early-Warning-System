# 🩺 Aavishkar Patient Monitor - Exploratory Data Analysis (EDA)
## PhysioNet / Computing in Cardiology Challenge 2019 Dataset

### Step 1 to 3: Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import seaborn as sns

sns.set_theme(style="darkgrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 5)
```

### Step 4 & 5: Dataset Location & File Counting

```python
DATA_PATH = Path("..") / "data" / "raw" / "physionet" / "training_setA"
files = list(DATA_PATH.glob("*.psv"))

print("Total patient files:", len(files))
```

### Step 6: First Patient Record Preview

```python
patient = pd.read_csv(
    files[0],
    sep="|"
)

patient.head()
```

### Step 7: Inspect Columns

```python
patient.columns
```

### Step 8: Dataset Information & Data Types

```python
patient.info()
```

### Step 9: Basic Vital Sign Statistics

```python
patient.describe()
```

### Step 10: Missing Values per Column

```python
patient.isnull().sum()
```

### Step 11: Missing Value Percentage

```python
missing = patient.isnull().mean() * 100
missing.sort_values(ascending=False)
```

### Step 12: Heart Rate (HR) Distribution

```python
patient["HR"].hist(
    bins=20
)

plt.title("Heart Rate Distribution")
plt.xlabel("BPM")
plt.ylabel("Hourly Count")
plt.show()
```

### Step 13: Oxygen (SpO2 / O2Sat) Distribution

```python
patient["O2Sat"].hist(
    bins=20
)

plt.title("SpO2 Distribution")
plt.xlabel("Percentage (%)")
plt.ylabel("Hourly Count")
plt.show()
```
